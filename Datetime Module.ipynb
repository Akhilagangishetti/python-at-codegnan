```python
from datetime import datetime, timedelta
curr_timestamp = datetime.now() #to get current timestamp
print(curr_timestamp)
today = (curr_timestamp.date()) #It extracts date from timestamp
curr_time = (curr_timestamp.time()) # it extract time
print(today)
print(curr_time)
print(today.year)
print(today.month)
print(today.day)
print(curr_time.hour, curr_time.minute, curr_time.second)
print(today.strftime("%d-%M-%Y"))
print(curr_time.strftime("%I-%M-%p"))
print(curr_time.strftime("%H-%M-%p"))
```

```python
import time

sec = 0
minute = 0
hour = 0

while True:
    print(f"{hour:02d}:{minute:02d}:{sec:02d}")

    time.sleep(1)

    sec += 1

    if sec == 60:
        sec = 0
        minute += 1

    if minute == 60:
        minute = 0
        hour += 1
```

```python
import time
from IPython.display import clear_output
sec_counter = min_counter = hour_counter = 0
while True:
    print(f"{hour_counter}:{min_counter}:{sec_counter}")
    time.sleep(1)
    sec_counter +=1
    if sec_counter == 60:
        min_counter += 1
        sec_counter =0
    if min_counter == 60:
        hour_counter += 1
        min_counter = 0
        if hour_counter == 24:
            hour_counter = 0
    clear_output(wait=True)
```

```python
import time
from IPython.display import clear_output
minutes = int(input("Enter stop watch minutes:"))
sec_counter = min_counter = hour_counter = 0
while min_counter != minutes:
    print(f"{hour_counter}:{min_counter}:{sec_counter}")
    time.sleep(1)
    sec_counter +=1
    if sec_counter == 60:
        min_counter += 1
        sec_counter =0
    if min_counter == 60:
        hour_counter += 1
        min_counter = 0
        if hour_counter == 24:
            hour_counter = 0
    clear_output(wait=True)
```
