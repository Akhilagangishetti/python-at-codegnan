```python
# List concatination
l1 = [1,2,3,4]
l2 = [4,5,6]
res = l1 +l2
print(res)
print(l1 + l1)
# List Reperatation
print(l1 *3)
print([0] * 10)
print([]*10)
```

```python
#### List Methods
 - Element Adding Methods
   - append(item): It adds item at end of the list(i.e, only single element at end)
   - extend(itarable): It adds all itarable elements at end of the list
   - insert(pos, item): It method adds the item at specified position 

[]:
```

```python
lst = [1,2,3,4]
lst.append(10)
print(lst)
lst.append('code')
print(lst)
lst.append([10,20,30])
print(lst)
print(lst[6])
lst.extend([100,200,300])
print(lst)
lst.insert(1,2000)
print(lst)
```

```python
- Element remove methods
  - remove(val): It removes the first occurance of value if value exists in list, otherwisw it raises 'ValueError'
  - pop(): It remove and returns the last of element of list
  - pop(pos): It remove and returns the last of element of list
  - clear(): It removes all element from list
  - del: It python keyword, used to delete values.

[]:
```

```python
lst = [10,20,30,20,40]
lst.remove(20)
print(lst)
# lst.remove(100) #ValueError: list.remove(x): x not in list
val = lst.pop()
print(val, lst)
val2 = lst.pop(1) #removes the 1st position element
print(val2, lst)
# print(lst.pop(10)) # IndexError: pop index out of range
del lst[0]
print(lst)
lst.clear()
print(lst)
```

```python
# index(val): It returns the index of first occurence of value if exists in list otherwise ValueError
# count(val): It returns the total frequence of value
lst = [11,2,12,4,14,2,2,2]
ind = lst.index(2)
print(ind)
print(lst.count(2))
print(lst.count(100))
print(lst.index(100))
```

```python
# sort(reverse =False) : It sorts the all list elements in ascending order, If reverse value is True it sorts in Descending order.
lst = [5,43,63,6,73,-2, 4,-6]
lst.sort()
print(lst)
```

```python
lst = [5,43,63,6,73,-2, 4,-6]
lst.sort(reverse=True)
print(lst)
```
