```python
class Account:

    def __init__(self, name, balance, password):
        self.name = name             # Public
        self._balance = balance      # Protected
        self.__password = password   # Private

    # Getter method
    def get_balance(self):
        return self._balance

    # Withdraw method
    def withdraw(self, amount):
        if self._balance >= amount:
            self._balance -= amount
            return f"{amount} withdrawn successfully"
        return "Insufficient Balance!"

    # Deposit method
    def deposit(self, amount):
        self._balance += amount
        return f"{amount} deposited successfully. Current Balance: {self._balance}"

    # Private method
    def __total_users(self):
        return "Total users: 100"

    # Setter method
    def update_password(self, new_password):
        self.__password = new_password
        return "Password updated successfully"

    # Getter method
    def get_password(self):
        return self.__password


# Object creation
user = Account("Akhila", 5000, "Akhila123")

print(user.name)
print(user.get_balance())
print(user.deposit(5000))

print(user.update_password("akki123"))
print(user.get_password())

# access the protected member syntax: obj._attributename
print(user._balance)
# access private member by using name mangling
print(user._Account__total_users())
```
