```python
# check weather the number is even or odd
num = int(input())
if num %2 == 0:
    print("Even")
if num %2 != 0:
    print("Odd")
```

```python
# check weather the number is even or odd
num = int(input())
if num %2 == 0:
    print(f' {num} num is Even')
if num %2 != 0:
    print(f' {num} num is Odd')
```

```python
num = int(input())
if num %3 == 0:
    print("Fizz")
if num %5 == 0:
    print("Buzz")
else:
    print(num)
```

```python
num = int(input())
if num %3 == 0:
    print("Fizz")
elif num %5 == 0:
    print("Buzz")
elif num %3 == 0 and num %5 == 0:
    print("Fizz and Buzz")
else:
    print(num)
```
