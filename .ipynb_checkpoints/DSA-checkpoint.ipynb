```python
from collections import Counter
string = "Nandhini"
c = Counter(string)
print(c)
print(c.most_common(2))
c.update("nnd")
print(c)
c.subtract("nnd")
print(c)
```

```python
s = "Akhilashetty"
rev = ""
for ch in s:
    rev = ch+rev
    print(rev)
    
if rev == s:
    print("It is a palindrome")
else:
    print("Not a palindrome")
    
```
