```python
# Built in functions: The Functions defination is provided by python, those functions called as Built in Functions
```

```python
#abs(): It returns the abuslute value
print(abs(20-10))
print(abs(10-20))
```

```python
# round(val, pos): This function rounds the float value to particular positions
print(round(2.2589, 2))
print(round(2.2539, 2))
print(round(2.2539))
print(int(2.2539))
```

```python
#ord(char): It returns ACSII value of character
# chr(ascii_val): It returns the character for specified ASCII value
print(ord("A"))
print(chr(65))
```

```python
# sorted(iterable, reverse=False, key=None): It return new sorted iterable
# sort the fruits based on the length of fruit name
fruits = ['Apple', 'banana', 'guava', 'papaya', 'grapes']
res = sorted(fruits)
print(res)
print(sorted(fruits, reverse = True))
print(sorted(fruits, reverse = True, key = len)) # it sorts the list based on length of each item/element length
```

```python
# map(function, iterable)
# Find the square of all list of integers using map function
def square(X):
    return X**2
arr = [1,2,3,4,5]
res = list(map(square, arr))
print(res)
# filter(function, iterable)
# find the even numbers from list of numbers
lst = [1,2,3,4,5,6]
```

```python
# filter(function, iterable)
# find the even numbers from list of numbers
def check_even(X):
    if X % 2 ==0:
        return X
lst = [1,2,3,4,5,6]
res = filter(check_even, lst)
print(res)
```

```python
# filter(function, iterable)
# find the even numbers from list of numbers
def check_even(X):
    if X % 2 ==0:
        return X
lst = [1,2,3,4,5,6]
res = list(filter(check_even, lst))
print(res)
```

```python
# filter(function, iterable)
# find the even numbers from list of numbers
def check_even(X):
    if X % 2 ==0:
        return 1
lst = [1,2,3,4,5,6]
res = list(map(check_even, lst))
print(res)
```

```python
# filter(function, iterable): It filters the iterable, if the function returns Non-None those values will added to result
# find the even numbers from list of numbers
def check_even(X):
    if X % 2 ==0:
        return X
lst = [1,2,3,4,5,6]
res = list(map(check_even, lst))
print(res)
```

```python
# find the 3's multipliers list of integers
arr = [1,24,54,22,44,51,33,15]
def Threes_multipliers(X):
    if X%3 ==0:
        return X
res = list(filter(Threes_multipliers, arr))
print(res)
```

```python
#enumerate(iterable): This function returns the list of tuple of index and item for iterable items/elements
fruits = ['Apple', 'banana', 'guava', 'papaya', 'grapes']
print(enumerate(fruits))
print(list(enumerate(fruits)))
for ind, val in enumerate(fruits):
    print(ind, val,len(val))
```

```python
#enumerate(iterable): This function returns the list of tuple of index and item for iterable items/elements
fruits = ['Apple', 'banana', 'guava', 'papaya', 'grapes']
print(enumerate(fruits))
print(list(enumerate(fruits)))
for ind, val in enumerate(fruits):
    print(f"{ind}-{val} and length is {len(val)}")
```

```python
# Tuple unpacking
t = (1,2)
a, b = t
print(a, b)
```

```python
# Zip(iterable1, iterable2, ...): It returns all iterables as single iterable of tuples
roll_no = [101,102,103,104,105]
names = ['Akki', 'Srav', 'Likith Kumar', 'Rithu', 'Manu']
percentage = [90,89,99.99999,96.8,87.65]
print(list(zip(roll_no, names, percentage)))
```

```python
num = int(input())
res = ""
while num > 0:
    bit = num % 2
    num = num // 2
    res = str(bit) + res
print(res)
```

```python
# number conversions
print(bin(10)) # string type of binary value
print(bin(10)[2:])
print(oct(10))
print(hex(11))
print(hex(11).upper()[2:])
```

```python
num=int(input())
res = ""
while num>0:
    bit = num % 16
    num = num // 16
    res = str(bit)+res
print(res)
```

```python
# any(): This Function returns True when any one iterable items return True
# all(): Return True if bool(x) is True for all values x in the iterable
arr = [1,2,3,4,0]
print(all(arr))
print(any(arr))
```

```python
percentage = [34,35,54,65,67]
for num in percentage:
    if num < 35:
        print("All students not passes")
        break
else:
        print("All students pass")
```

```python
percentage = [34,35,54,65,67]
lst = []
for num in percentage:
    if num < 35:
        lst.append(False)
    else:
        lst.append(True)
print(lst)
if all(lst):
    print("All students pass")
else:
    print("All students not passed")
```
