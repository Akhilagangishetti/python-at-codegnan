```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def details(self):
        return f"My name is {self.name} and I am a {self.age} years old"
class Student(Person):
    def __init__(self,name,age,rollno):
        super().__init__(name, age)
        self.rollno = rollno
    def details(self):
        return f"I am student"
p = Person("Akki", 15)
s = Student("Akkki", 15, 1001)
print(p.details())
print(s.details())   
```

```python
# Method overriding
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    def details(self):
        return f"My name is {self.name} and I am a {self.age} years old"
class Student(Person):
    def __init__(self,name,age,rollno):
        super().__init__(name, age)
        self.rollno = rollno
    def details(self):
        return super().details() + f" I am student"
p = Person("Akki", 15)
s = Student("Akkki", 22, 1001)
#print(p.details())
print(s.details())   
```

```python
class math:
    def add(self, x, y):
        return x + y
    def add(self, x,y,z):
         return x + y + z
    def add(self, x,y,z,a):
         return x + y + z + a
m = math()
print(m.add(1,2,3,4))
print(m.add(1,2,3))
print(m.add(1,2))
    
    
```

```python
# Python directly does not support method overloading, but we can achieve through the default arguments or orbitary arguments
```

```python
# Method overloading
class Math:
    
    def add(self, a,b, c=0, d=0):
         return a + b + c + d
m = Math()
print(m.add(1,2,3,4))
print(m.add(1,2,3))
print(m.add(1,2))
```

```python
# Operater Overloading
class point:
    def __init__(self, x):
        self.x = x
    def __str__(self): # String magic method
        return f" This is point class and value is:({self.x})"
p1 = point(5,3)
print(p1)
```

```python
# here adding for 2 objects
class point:
    def __init__(self, x):
        self.x = x
    def __str__(self): # String magic method
        return f" This is point class and value is:({self.x})"
    def __add__(self, other):
        return self.x + other.x
p1 = point(5)
p2 = point(3)
print(p1.x + p2.x)
print(p1 + p2)
print(p1 - p2)
```

```python
# here adding for 2 objects
class point:
    def __init__(self, x):
        self.x = x
    def __str__(self): # String magic method
        return f" This is point class and value is:({self.x})"
    def __add__(self, other):
        return self.x + other.x
    def __sub__(self, other):
        return self.x - other.x
p1 = point(5)
p2 = point(3)
print(p1.x + p2.x)
print(p1 + p2)
print(p1 - p2)
```

```python
# here adding for 2 objects
class point:
    def __init__(self, x):
        self.x = x
    def __str__(self): # String magic method
        return f" This is point class and value is:({self.x})"
    def __add__(self, other):
        return str(self.x) + str(other.x)
    def __sub__(self, other):
        return self.x - other.x
p1 = point(5)
p2 = point(3)
print(p1.x + p2.x)
print(p1 + p2)
print(p1 - p2)
```
