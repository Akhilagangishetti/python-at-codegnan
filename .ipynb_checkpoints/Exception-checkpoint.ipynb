```python
# Exception handling block
# without exception
try:
    a = int(input())
    b = int(input())
    c = a + b
    print(c)
except:
    print("Something wrong in try block")
else:
    print("There is no exception")
finally:
    print("Program done")
```

```python
# Exception handling block
# with an exception
try:
    a = int(input())
    b = int(input())
    c = a + b
    print(c)
except:
    print("Something wrong in try block")
else:
    print("There is no exception")
finally:
    print("Program done")
```

```python
a = 10
b = 0
c = a / b
print(c)
```

```python
# Exception handling block
# Exception with exception class
try:
    a = 10
    b = 0
    c = a / b
    print(c)
except Exception as e:
    print(f"Something wrong in try block:{e}")
```

```python
# Exception handling block
# Exception with exception class
try:
    a = 10
    b = 0
    c = a.append(100)
    print(c)
except Exception as e:
    print(f"Something wrong in try block:{e}")
```

```python
# Exception handling block
# Define multiple exception with different except block
try:
    a = 10
    b = 0
    c = a / b
    print(c)
except ZeroDivisionError as e:
    print(f" Zero division not possible: {e}")
except IndexError as e:
    print(f" We can't access element at this position: {e}")
except Exception as e:
    print(f"Something wrong in try block:{e}")
```

```python
# Exception handling block
# Define multiple exception with different except block
try:
    arr = [1,2,3,4]
    c = arr[10]
    print(c)
except ZeroDivisionError as e:
    print(f" Zero division not possible: {e}")
except IndexError as e:
    print(f" We can't access element at this position: {e}")
except Exception as e:
    print(f"Something wrong in try block:{e}")
```

```python
# Exception handling block
# Define multiple exception with different except block
try:
    a = 10
    c = len(a)
    print(c)
except ZeroDivisionError as e:
    print(f" Zero division not possible: {e}")
except IndexError as e:
    print(f" We can't access element at this position: {e}")
except Exception as e:
    print(f"Something wrong in try block:{e}")

```
