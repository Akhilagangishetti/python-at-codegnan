```python
import re
```

```python
# match(pat, string)
string = "The sky is blue and Ocean is blue"
res = re.match("[A-Z]", string)
print(res)
print(res.span()) # it returns the matched positions
print(res.group()) # it returns matched pattern val
```

```python
# match(pat, string)
string = "The sky is blue and Ocean is blue"
res = re.match("[A-Za-z]+", string)
print(res)
print(res.span()) # it returns the matched positions
print(res.group()) # it returns matched pattern val
```

```python
# match(pat, string)
try:
    string = "The sky is blue and Ocean is blue"
    res = re.match("[z]+", string)
    print(res)
    print(res.span()) # it returns the matched positions
    print(res.group()) # it returns matched pattern val
except Exception as e:
    print("There is no pattern matched:", e)
    
```

```python
# search(pat, string)
# search 5 length word
import re
try:
    string = "The sky is blue and Ocean is blue"
    res = re.search("[a-zA-Z]{5}", string)
    print(res)
    print(res.span()) # it returns the matched positions
    print(res.group()) # it returns matched pattern val
except Exception as e:
    print("There is no pattern matched:", e)
    
```

```python
#get the words from string which length is morethan 3
# findall(pat,string)
import re
try:
    string = "The sky is blue and Ocean is blue"
    res = re.findall("[a-zA-Z]{3,}", string)
    print(res)
    
except Exception as e:
    print("There is no pattern matched:", e)
    
```

```python
import re
try:
    string = "The sky is blue and Oce456an is b123lue"
    res = re.findall("[a-zA-Z]+", string)
    print(res)
    
except Exception as e:
    print("There is no pattern matched:", e)
    
```

```python
# Validate mobile number
# All are digits, length of number is 10, starts with 6, 7,8,9, Remaning are 0-9
import re
try:
    mobile = input()
    res = re.findall("^[6-9][0-9]{9}", mobile)
    print(res)
    if res:
        print(f"{mobile} is valid")
    else:
        print(f"{mobile} is  Invalid")
    
except Exception as e:
    print("There is no pattern matched:", e)
    
```

```python
# PAN  number
# it allows upper and digits, Length is 10, first 
import re
try:
    pan = input()
    res = re.findall("^[A-Z]{5}[0-9]{4}[A-Z]$", pan)
    print(res)
    if res:
        print(f"{pan} is valid")
    else:
        print(f"{pan} is  Invalid")
except Exception as e:
    print("There is no pattern matched:", e)
```

```python
# Validation of Gmail
# it allows lower and digits, Length
import re
try:
    email = input()
    res = re.findall(r"^[a-z0-9][a-z0-9.]+\@[a-z0-9.]+\.[a-z]{2,3}", email)
    print(res)
    if res:
        print(f"{email} is valid")
    else:
        print(f"{email} is  Invalid")
except Exception as e:
    print("There is no pattern matched:", e)
```

```python
# Validation of Gmail
# it allows lower and digits, Length
import re
try:
    email = input()
    res = re.findall(r"^[a-z0-9][a-z0-9.]+\@[a-z0-9.]+\.[a-z]{4,5}", email)
    print(res)
    if res:
        print(f"{email} is valid")
    else:
        print(f"{email} is  Invalid")
except Exception as e:
    print("There is no pattern matched:", e)
```

```python
# Validate Date
import re
try:
    date = input()
    res = re.fullmatch(r"^(0?[1-9]|[12][0-9]|3[01])[-/](0?[1-9]|1[0-2])[-/]([0-9]){4}", date)
    print(res)
    if res:
        print(f"{date} is valid")
    else:
        print(f"{date} is  Invalid")
    
except Exception as e:
    print("There is no pattern matched:", e)
    
```
