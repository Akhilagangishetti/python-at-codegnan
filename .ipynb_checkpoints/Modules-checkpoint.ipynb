```python
import math
print(dir(math))
```

```python
print(math.ceil(4.956)) # It returns the upper bound value
print(math.floor(4.25)) # It returns the lower bound value
print(math.factorial(5)) # It returns factorial of a number
print(math.gcd(10,5,2)) # It returns the GCD value
print(math.lcm(45,5))  # It returns the LCM value
print(math.log10(2))
print(math.log10(4))
print(math.log2(4))
print(math.pi)
```

```python
import random
print(random.random()) # This Function returns the value in between 0 to 1
print(random.randint(1,10)) # randint(a,b): This Function generates random number between 'a' and 'b'
print(random.choice(['A','K', 1,2,3,4,5,6,7,8,9, 'J', 'Q']))
print(random.choices(['A','K', 1,2,3,4,5,6,7,8,9, 'J', 'Q'], k=3))
arr = ['A','K', 1,2,3,4,5,6,7,8,9, 'J', 'Q']
random.shuffle(arr)
print(arr)
```

```python
# who will reach top in less steps
import random 
steps = rolls = 0
while True:
    input("To roll dice for that click enter button")
    dice_val = random.randint(1,6)
    rolls += 1
    if steps + dice_val <= 100:
        steps += dice_val
    if steps == 100:
        print(f"Your reached TOP in {rolls} rolls")
        break
    print(f" For {rolls} rolls your at {steps} position")
```
