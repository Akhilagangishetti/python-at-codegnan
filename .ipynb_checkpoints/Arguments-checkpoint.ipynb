```python
# Types of Arguments
# 1. Positional Arguments: The arguments are assigns to parameters as per the positional order
def sub(x:int, y:int) ->int:
    c = x - y
    return c
#function calling block
a, b = 10, 20
res = sub(a, b)
print(res)
print(sub(b, a))
```

```python
# Types of Arguments
# 1. Keyword Arguments: Here arguments are passed with along with the parameters
def sub(x:int, y:int) ->int:
    c = x - y
    return c
#function calling block
a, b = 10, 20
print(sub(x = a, y = b))
print(sub(y = b, x = a))
```

```python
def add(a, b):
    c = a + b
    return c
#function calling block
a,b = 5,2
res = add(a,b)
print(res)

    
```

```python
def add(a,b,c):
    x = a + b + c
    return x
#function calling block
a,b,c = 1,6,4
res = add(a,b,c)
print(res)
```

```python
# Default arguments: The parameters defined with some default values, and it assign the argument value f you pass the values for default parameters otherwise, it will assign default values.
def add(a,b,c = 0 , d = 0):
    res = a+b+c+d
    return res
print(add(10,20))
print(add(10,20, 5))
print(add(10,20, 2, 1))
```

```python
def mul(a,b, c = 1, d = 1):
    res = a*b*c*d
    return res
print(mul(10,30))
print(mul(10,20, 5))
print(mul(10,20, 2, 1))
```

```python
# 4. variable length of positional arguments
def add(a, b = 0 , *args):
    print(type(args))
    print(args)
print(add(1,2,3,4,5,6,7,8,9))
```

```python
# 4. variable length of positional arguments
def add(a, b = 0 , *args):
    print(type(args))
    print(args)
    s = 0
    for num in args:
        s += num
    return s
print(add(1))
# print(add(1, a=1, b=2))
print(add(1,2,3,4,5,6,7,8,9))
```

```python
# 4. variable length of positional arguments
def add(a, b = 0 , *args):
    print(type(args))
    print(args)
    s = 0
    for num in args:
        s += num
    return s + a + b 
print(add(1))
# print(add(1, a=1, b=2))
print(add(1,2,3,4,5,6,7,8,9))
```

```python
#variable length of keywords arguments
def details(name, age, *marks, **kwargs):
    print(type(kwargs))
    print(kwargs)
print(details("Ravi",  20, 85,95,75, gender="male", location="India"))
```

```python
#variable length of keywords arguments
def details(name, age, *marks, **kwargs):
    print(type(kwargs))
    print(kwargs)
    print(kwargs.keys())
print(details("Ravi",  20, 85,95,75, gender="male", location="India"))
```

```python
# pass by value
def fun(x):
    print(f" Before update Address of x :{id(x)} and value: {x}")
    x = x + 1
    print(f"After update the x value, Address of x :{id(x)} and value: {x} ")
a = 5
print(f"Before Function Execution address of a:(id({a}) and value {a}")
print(fun(a))
print(f" After Function Execution address of a:(id({a}) and value {a}")
```

```python
# pass by reference
def fun(x):
    print(f"Before update Address of x :{id(x)} and value: {x}")
    x.append(100)
    print(f"After update the x value, Address of x :{id(x)} and value: {x} ")
a = [1,2,3]
print(f"Before Function Execution address of a:(id({a}) and value {a}")
print(fun(a))
print(f" After Function Execution address of a:(id({a}) and value {a}")
```
