```python
#### Set Methods
- Adding Element To Set
  - add(ele): It adds element to set if element not exists in set, otherwise nothing
  - update(iterable): It adds all iteable elements to set
- Remove elements from Set
  - remove(ele): It removes the element from set if element exists in set otherwise
'KeyError'
  - discard(ele): It removes the element from set if element exists in set otherwise nothing
  - pop(): It removes and returns the element from set
-Set Operations
  - union(another_sets): It returns the all commom elements from all sets
  - intersection(amother_set): It returns the common elements from the all sets
  - difference(another_sets): It returns these set elements those are not lies in another_sets
  - symmetric_difference(another_sets): It returns the Non-common elements from all sets
```

```python
a = {2,3,5,7,9}
a.add(1)
print(a)
a.add(2) #2 doesn't add to set because set contains 2
print(a)
a.remove(9)
print(a)
a.update([10,40])
print(a)
#a.remove(9)
#print(a) # KeyError
a.discard(7)
print(a)
a.pop()
print(a)
```

```python
a1 = {1,3,4,7,9}
a2 = {2,5,3,8,7}
a3 = {6,9}
print(a1.union(a2))
print(a1.union(a2, a3))
print(a1.intersection(a2))
print(a1.intersection(a2, a3))
print(a1.difference(a2))
print(a1.difference(a2, a3))
print(a1.symmetric_difference(a2))
print(a1.symmetric_difference(a2, a3)) #set.symmetric_difference() takes exactly one argument (2 given)
```

```python
a1 = {1,3,4,7,9}
a2 = {2,5,3,8,7}
a3 = {6,9}
print(a1.union(a2))
print(a1.union(a2, a3))
print(a1.intersection(a2))
print(a1.intersection(a2, a3))
print(a1.difference(a2))
print(a1.difference(a2, a3))
print(a1.symmetric_difference(a2))
print(a1.symmetric_difference(a3))
```

```python
a1 = {1,3,4,7,9}
a2 = {2,5,3,8,7}
a3 = {6,9}
print(a1 | a2) # union
print(a1 | a2 | a3)
print(a1 & a2) # intersection
print(a1 & a2 & a3)
print(a1 - a2) # for difference
print(a1 - a2 - a3)
print(a1 ^ a2) #symmetric difference
print(a1 ^ a3)
```
