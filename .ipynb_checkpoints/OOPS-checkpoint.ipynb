```python
# create mobile class
class Mobile:
    company = "Iphone" # class attribute
    def __init__(self, model, ram, rom, price):
        self.model = model
        self.ram = ram
        self.rom = rom
        self.price = price
# object creation
iphone16 = Mobile("iphone 16", 16, 256, 125000)
iphone17 = Mobile("iphone 17", 8, 256, 120000)
print(iphone16.price)
print(iphone17.price)
print(iphone16.company)
print(iphone17.company)
print(Mobile.company)
```

```python
# updating Instance attribute data from outside class
class Mobile:
    company = "Iphone" # class attribute
    def __init__(self, model, price, ram = 8, rom = 256):
        self.model = model
        self.ram = ram
        self.rom = rom
        self.price = price
# object creation
iphone16 = Mobile("iphone 16", 80000)
iphone17 = Mobile("iphone 17", 120000, 16)
print(iphone16.ram)
iphone16.ram = 16
print(iphone16.ram)
iphone16.company = "New Iphone"
```

```python
# create mobile class
class Mobile:
    company = "Iphone" # class attribute
    def __init__(self, model, ram, rom, price):
        self.model = model
        self.ram = ram
        self.rom = rom
        self.price = price
    def about_phone(self):
        return f"The {self.model} have {self.ram}GB RAM and {self.rom}GB ROM"
    def display():
        return "This is samsung company display"
    # class method n 
    @classmethod
    def update_company(cls, new_name):
        cls.company - new_name
        return "Company name updated"
    # static method
    @staticmethod
    def charger_provide(cable = False):
        return "adpter + cable" if cable else "only adpter"
        
# object creation
iphone16 = Mobile("iphone 16", 16, 256, 125000)
iphone17 = Mobile("iphone 17", 8, 256, 120000)
#print(iphone.about_phone())
print(iphone16.update_company("Iphone Pro"))
print(iphone16.charger_provide())
print(iphone17.company)
```

```python
class Account:
    Bank = "Akhila"
    def __init__(self, username, balance, password):
        self.username = username
        self.balance = balance
        self.password = password
    def withdraw(self,amount):
        if self.balance >= amount:
            self.balance -= amount
            return f"{amount} withdrawal successfully"
        return "Insufficient Balance"
    def deposite(self, amount):
        self.balance += amount
        return f"{amount} Deposite successfully and current Balance is {self.balance}"
user = Account("Akhila", 10000, "akkki@123")
print(user.deposite(1000))
user.balance = 100000
print(user.withdraw(50000))   
```
