```python
s = "B1ACEF2354"
upper = 0
digits = 0
visited = set()
valid = True
if len(s) != 10:
    valid = False
    
for ch in s:
    if not ch.isalnum():
        valid = False
        break
    if ch in visited:
        valid = False
    visited.add(ch)
    if ch.isupper():
        upper+=1
    elif ch.isdigit():
        digits+=1
if valid and upper>2 and digits>3:
    print("valid")
else:
    print("Invalid")
```
