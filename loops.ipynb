```python
# Remove target element from list of integers
arr = list(map(int, input().split()))
target = int(input())
res = []
for num in arr:
    if num != target:
        res.append(num)

print(res)
```

```python
arr = list(map(int, input().split()))
target = int(input())
res = []
n = len(arr)
i = 0 # initilization
while i < n:
    if arr[i] != target:
        res.append(arr[i])
    i+= 1

print(res)
```

```python
arr = list(map(int, input().split()))
target = int(input())

n = len(arr)
i = 0 # initilization
while i < n:
    if arr[i] == target:
        arr.pop(i)
        n-=1
    else:
        i+= 1

print(arr)
```

```python
# return the target element position if target exists otherwise -1
arr = list(map(int, input().split()))
target = int(input())
res = -1
n = len(arr)
for i in range(n):
    if arr[i] == target:
        res = i
print(res)
```
