```python
# sum of two numbers uding lambda
add = lambda x, y : x + y
print(add(3, 4))
```

```python
# find square of all list of integer using lambda and map()
arr = [1,2,3,4,5,6]
res = list(map(lambda x : x**2, arr))
print(res)
```

```python
# find the even numbers in list using lambda and filter()
arr = [1,2,3,4,5,6]
res = list(filter(lambda x : x%2==0, arr))
print(res)
```

```python
# find the weather the number is even or odd using lambda
n = int(input())
even_or_odd = lambda x:"odd" if x%2 else "even" 
print(even_or_odd(n))
```

```python
# find the weather the number is even or odd using lambda
n = int(input())
even_or_odd = lambda x:f"{x} - odd" if x%2 else f"{x} -  even" 
print(even_or_odd(n))
```
