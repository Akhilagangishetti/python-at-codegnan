```python
age = int(input())
health_score = int(input())
vehicle_type = input().strip()

premium = 1000
#Age Factor 
if age < 25:
    premium = premium*1.20
elif age > 50:
    premium = premium * 1.15

# Health_score factor 
if health_score >= 80:
    premium = premium*0.90
elif health_score < 60:
    premium = premium*1.20

#Vehicle type factor 
if vehicle_type == "Sports" or vehicle_type == "Sports Car":
    premium = premium*1.30
elif vehicle_type == "SUV":
    premium = premium*1.15

print(float(premium))
```
