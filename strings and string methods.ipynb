### String Methods
- Case Conversion Methods
  - upper(): it converts entire into uppercase
  - lower(): it converts entire into lowercase
  - title(): it converts every word first character into uppercase and remaining into lowercase
  - swapcase(): it converts lower into upper and upper into lower
  - capitalize(): it converts first character of string into uppercase and remaining into lowercase

[]:

```python
string = "The ocean is bule and sky is BLUE 123442@"
lower = string.lower()
print(lower)
print(string.upper())
print(string.title())
print(string.swapcase())
print(string.capitalize())
```

```python
# White space remove methods
# strip(): it removes the white space to both sides of string
#lstrip(): it removes the white space at left/leading side of string
#rstrip(): it removes the white space at right/lagging side of string
string ="        codegnan        "
print(string)
print(string.strip())
print(string.lstrip())
print(string.rstrip())
```

```python
# White space remove methods
# strip(): it removes the white space to both sides of string
#lstrip(): it removes the white space at left/leading side of string
#rstrip(): it removes the white space at right/lagging side of string
string =" @@@@@@@    code  @ gnan      @@@@@@"
print(string)
print(string.strip("@"))
print(string.lstrip("@"))
print(string.rstrip())
```

```python
# split():It returns the list of substring and by default splits at white space
string = "The ocean is bule and sky is BLUE 123442@"
print(string.split())
print(string.split('is'))
print(string.split('z'))
print(string.split('blue'))
print(string.split('@'))
```

```python
# Indexing finding methods
# index(sub_string): It returns the first occurance index of sub string if exist, otherwise "valueError"
# find(sub_string): It returns the first occurance index of sub string if exist, otherwise "-1"
string = "The ocean is bule and sky is blue 123442@"
print(string.index("e"))
print(string.index("o"))
print(string.index("is"))
print(string.index('is', 12)) # it starts checking from 14th index
print(string.index('z'))
```

```python
# Indexing finding methods
# index(sub_string): It returns the first occurance index of sub string if exist, otherwise "valueError"
# find(sub_string): It returns the first occurance index of sub string if exist, otherwise "-1"
string = "The ocean is bule and sky is blue 123442@"
print(string.find("e"))
print(string.find("o"))
print(string.find("is"))
print(string.find('is', 12)) # it starts checking from 14th index
print(string.find('z'))
```

```python
# replace(old, new): It replace the old_string with new_string in string
string = "The ocean is bule and sky is blue 123442@"
print(string.replace('blue', 'red'))
print(string.replace('blue', 'red',1)) # It replaces only 1 occurance
```

```python
# "sep".join(iterable): It joins the all iterable elements with seperator
lst = ['The', 'ocean', 'is', 'bule', 'and', 'sky', 'is', 'BLUE']
print("".join(lst))
print(" ".join(lst))
print("_".join(lst))
```

```python
#### string checking methods
- islower(): It returns True when entire string is in lowercase
- isupper(): It returns True when entire string is in uppercase
- istitle(): It returns True when string is title format
- isalpha(): It returns True when string contains alphabets(i.e upper & lowercase)
- isalnum(): It returns True when string contains alaphabets or digits
- isdigit(): It returns True when string contains only digits
- startswith(substring): It returns true string starts with substring
- endswith(substring): It returns true string ends with substring

[]:

```

```python
string = "The ocean is bule and sky is blue 123442@"
print(string.islower())
print(string.isupper())
print(string.istitle())
print(string.isalpha())
print(string.isalnum())
print(string.isdigit())
print(string.startswith('o'))
print(string.endswith('b'))
```
