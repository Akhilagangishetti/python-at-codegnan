```python
# reading integer from user
# integer to float, string, bool
n = int(input())
print(type(n), n)
print(float(n))
print(str(n))
print(bool(n))
```

```python
# reading float value from user
# float to int, string, bool
val = float(input())
print(type(val), val)
print(int(val))
print(str(val))
print(str(val))
print(bool(val))
```

```python
# reading complex value
real = int(input())
img = int(input())
c = complex(real, img)
print(type(c), c)
```

```python
# reading string value from user
s = input() # "123"
print(type(s), s)
print(int(s))
print(float(s))
print(list(s)) # It returns list of characters
print(tuple(s)) # It returns tuple of characters
print(set(s)) # It returns set of characters
```

```python
# reading string value from user
s = input() # "hello"
print(type(s), s)
# print(int(s))
l = list(s)
print(l[:-4:5]) # It returns list of characters
print(tuple(s)) # It returns tuple of characters
print(set(s)) # It returns set of characters
```

```python
# Reading Tuple of intergers
t = tuple(map(int, input().split()))
print(list(t))
print(set(t))
print(str(t))
```

```python
# Reading set of intergers
# Reading Tuple of intergers
a = set(map(int, input().split()))
print(list(a))
print(tuple(a))
print(str(a))
```
