```python
x, y = map(int, input().split())

print(x, "+", y, "=", x + y)
print(x, "*", y, "=", x * y)
print(x, "-", y, "=", x - y)
```

```python
A, B, C, D = map(int, input().split())

X = (A * B) - (C * D)

print(f"Difference = {X}")
```

```python
pi = 3.141592653

R = float(input())

area = pi * (R ** 2)

print(f"{area:.9f}")
```

```python
import math

A, B = map(int, input().split())

result = A / B

print(f"floor {A} / {B} = {math.floor(result)}")
print(f"ceil {A} / {B} = {math.ceil(result)}")
print(f"round {A} / {B} = {round(result)}")
```

```python
A, B = map(int, input().split())

if A >= B:
    print("Yes")
else:
    print("No")
```

```python
A, B = map(int, input().split())

if A % B == 0 or B % A == 0:
    print("Multiples")
else:
    print("No Multiples")
```
